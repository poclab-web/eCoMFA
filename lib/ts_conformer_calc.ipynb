# RDKit Conformer Check

This notebook builds a molecule from SMILES, generates multiple conformers, runs UFF optimization, prints basic diagnostics, and renders each conformer in `py3Dmol`.


```python
import os

from rdkit import Chem
from rdkit.Chem import AllChem, Lipinski

import py3Dmol


def calc(smiles):
    """Generate/optimize conformers from a SMILES string and visualize each conformer."""
    print(smiles)

    mol = Chem.MolFromSmiles(smiles)
    print(Lipinski.NumRotatableBonds(mol))

    mol = Chem.AddHs(mol)
    print(mol.GetSubstructMatch(Chem.MolFromSmarts("[#6](~[#8H0])([#6])([#6:1])")))

    fragments = Chem.GetMolFrags(mol, asMols=True)
    print(len(fragments))
    for frag in fragments:
        print(Chem.rdinchi.MolToInchiKey(frag))

    AllChem.EmbedMultipleConfs(mol, numThreads=0, numConfs=20, randomSeed=3)
    res = AllChem.UFFOptimizeMoleculeConfs(
        mol,
        numThreads=0,
        ignoreInterfragInteractions=False,
        maxIters=2000,
    )
    print(res)

    volume = Chem.rdMolDescriptors.DoubleCubicLatticeVolume(mol).GetVolume()
    print(volume)

    os.makedirs("../conformer", exist_ok=True)

    for conf, result in zip(mol.GetConformers(), res):
        print(result)
        block = Chem.rdmolfiles.MolToXYZBlock(mol, confId=conf.GetId())

        with open(f"../conformer/output{result[1]}.txt", "w") as f:
            print(block, file=f)

        viewer = py3Dmol.view(width=800, height=400)
        viewer.addModel(block, "mol")
        viewer.setStyle({"stick": {}})
        viewer.zoomTo()
        viewer.show()


# calc("c1ccccc1[C@@]0(C)O[B@-](Cl)([C@@H]2[C@@]([O]0)(C)[C@H](C3)C(C)(C)C3C2)([C@@H]4[C@@H](C)[C@H](C5)C(C)(C)C5C4)")
# calc("c1ccccc1[C@@]0(C)O[B-]4(C)OC(c2ccccc2)(c3ccccc3)[C@H]5CCC[N+]45[BH2-]O0")
calc("c1ccccc1[C@]0(C)O[B-]4([C@@H]2[C@@]([O]0)(C)[C@H](C3)C(C)(C)C3C2)([C@@H]5CCCC4CCC5)")
# calc("B4([C@@H]2[C@@H](C)[C@H](C3)C(C)(C)C3C2)([C@@H]5CCCC4CCC5)")
```
